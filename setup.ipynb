```python
complete_rows = []
raw_rows = []
mean_rows = []
for path in file_names:
    try:
        with open(path, 'rb') as src:
            obs = pickle.load(src)
        obs.reset_lst()
        obs.reset_sentinel_indexes()

        complete_row, raw_row, mean_row = complete_statistical_analysis(obs)

        complete_rows.append(complete_row)
        raw_rows.append(raw_row)
        mean_rows.appen(mean_row)

        # Save
        pd.DataFrame(complete_rows).to_csv('saved_observations/stats/complete_stats.csv', index=False)
        pd.DataFrame(raw_rows).to_csv('saved_observations/stats/per_pixel_stats.csv', index=False)
        pd.DataFrame(mean_rows).to_csv('saved_observations/stats/localized_mean_stats.csv', index=False)

        print(f'{obs.city_name.iloc[0]} iou stats collected')

    except Exception as e: print(f'{obs.city_name.iloc[0]} passed by')

```
